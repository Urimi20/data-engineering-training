```python

CATALOG = "ecommerce_catalog"
SCHEMA = "ecommerce_schema"

CUSTOMERS_PATH = f"/Volumes/{CATALOG}/{SCHEMA}/data/customers_raw.csv"
ORDERS_PATH = f"/Volumes/{CATALOG}/{SCHEMA}/data/orders_raw.csv"
customers_raw = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(CUSTOMERS_PATH)


customers_raw.display()

customers_raw.printSchema()
```

```python
orders_raw = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv(ORDERS_PATH)


orders_raw.display()

orders_raw.printSchema()
```

```python
print("Customers:", customers_raw.count())
print("Orders:", orders_raw.count())
```

```python
# raw to bronze

customers_raw.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "ecommerce_catalog.ecommerce_schema.bronze_customers"
    )


orders_raw.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "ecommerce_catalog.ecommerce_schema.bronze_orders"
    )
```
